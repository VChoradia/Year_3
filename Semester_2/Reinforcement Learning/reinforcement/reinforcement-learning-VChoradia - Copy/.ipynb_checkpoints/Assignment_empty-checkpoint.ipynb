```python
reward_lst = []

# Initialise the environment
# An arbitary gamma value is used here. You may need to change this!
env1 = get_task1_gridworld()

# Reset the environment to start from the starting location. 
# The reset() method returns the state, reward, and whether the episode is done or not
# More information can be found under the definition of the method
_, state, reward, done = env1.reset()
reward_lst.append(reward)

# While the trial is not done, i.e. the agent has not fallen into lava,
# the maximum length of the episode has not been exceeded, and the target has not been reached
while not done:
    
    # Take a step using the step() function and passing in one of the available actions [0, 1, 2, 3] 
    # corresponding to ['N', 'S', 'E', 'W']
    # More information can be found under the definition of the method
    # In this very simple example for each step we choose a random action
    _, state, reward, done = env1.step(np.random.choice(4))
    reward_lst.append(reward)

print(f'Collected rewards of an episode: {reward_lst}')
```

```python
def run_experiment(agent_class, env, gamma, alpha, epsilon, policy_type, n_episodes):
    agent = agent_class(env=env, policy_type=policy_type, alpha=alpha, gamma=gamma, episodes=n_episodes)
    if policy_type == 'softmax':
        _, total_rewards, _ = agent.solve(tau=1.0)
    else:
        _, total_rewards, _ = agent.solve(epsilon=epsilon)
    return np.convolve(total_rewards, np.ones(10)/10, mode='valid')

def plot_learning_curves(sarsa_rewards, q_learning_rewards, gamma, alpha, epsilon, policy_type):
    episodes = np.arange(len(sarsa_rewards))
    plt.figure(figsize=(10, 6))
    plt.plot(episodes, sarsa_rewards, label='SARSA')
    plt.plot(episodes, q_learning_rewards, label='Q-Learning')
    plt.xlabel('Episodes')
    plt.ylabel('Cumulative Rewards (Smoothed)')
    title = f'Learning Curve (γ={gamma}, α={alpha}, ε={epsilon}, policy={policy_type})'
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
# Direction mapping (N, E, S, W)
direction_map = {
    0: (0, 1),   # North
    1: (0, -1),  # South
    2: (1, 0),   # East
    3: (-1, 0)   # West
}

def plot_directions(gridworld, policy, title):
    plt.figure(figsize=(6, 6))
    grid = get_grid(gridworld)
    plt.imshow(grid)
    plt.xticks(fontsize=7)
    plt.yticks(fontsize=7)
    plt.grid(False)

    grid_shape = gridworld.get_gridshape()

    for row in range(grid_shape[0]):
        for col in range(grid_shape[1]):
            # Check if this location is valid
            if not gridworld._is_location((row, col)):
                continue

            state = gridworld._get_state_from_loc((row, col))

            # Check if the state is terminal (goal or lava) or an invalid cell (wall)
            if gridworld.is_terminal(state):
                continue

            # Get the preferred action from the policy
            action = policy[state]
            direction = direction_map.get(action)

            # Plot the arrow only if a valid direction is available
            if direction:
                dx, dy = direction
                plt.arrow(col, row, dx * 0.3, dy * 0.3, head_width=0.1, head_length=0.1, fc='blue', ec='blue')
    
    plt.title(title)
    plt.show()
```

# Task 2 - Dual Targets with Stochastic Rewards

```python
# This should be the definition of the gridworld for task 1. DO NOT CHANGE THIS
env2 = get_task2_gridworld()
```

```python
def softmax(x, tau):
    """Compute softmax values for each set of scores in x."""
    x = np.array(x) / tau
    x = x - np.max(x)  # For numerical stability
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)
```

```python
class QLearningSoftmax:
    def __init__(self, gridworld, tau=1.0, alpha=0.1, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau = tau
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.Q = np.zeros((self.gridworld.get_state_size(), self.gridworld.get_action_size()))

    def choose_action(self, state):
        action_probs = softmax(self.Q[state], self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)  # random seed for reproducibility
        episode_rewards = {goal: [] for goal in self.gridworld.get_goal_loc()}

        for _ in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            total_reward = 0

            while not done:
                action = self.choose_action(state)
                t, next_state, reward, done = self.gridworld.step(action)
                total_reward += reward

                best_next_action = np.argmax(self.Q[next_state])
                td_target = reward + self.gamma * self.Q[next_state, best_next_action]
                td_error = td_target - self.Q[state, action]
                self.Q[state, action] += self.alpha * td_error

                state = next_state
            
            # Track rewards for each goal location
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if state == goal_state:
                    episode_rewards[goal].append(total_reward)
        
        return self.Q, episode_rewards
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(x, tau):
    """Compute softmax values for each set of scores in x."""
    x = np.array(x) / tau
    x = x - np.max(x)  # For numerical stability
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

class QLearningSoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.1, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.Q = np.zeros((self.gridworld.get_state_size(), self.gridworld.get_action_size()))
        self.tau = tau_initial
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        action_probs = softmax(self.Q[state], self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(self.gridworld.get_action_size(), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)  # Set random seed for reproducibility
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            total_reward = 0

            while not done:
                action = self.choose_action(state)
                t, next_state, reward, done = self.gridworld.step(action)
                total_reward += reward

                best_next_action = np.argmax(self.Q[next_state])
                td_target = reward + self.gamma * self.Q[next_state, best_next_action]
                td_error = td_target - self.Q[state, action]
                self.Q[state, action] += self.alpha * td_error

                state = next_state
            
            # Track visits and rewards for each goal location
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if state == goal_state:
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            # Store data for plotting
            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            # Decay tau
            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.Q, goal_visits, cumulative_rewards

class SARSASoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.1, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.Q = np.zeros((self.gridworld.get_state_size(), self.gridworld.get_action_size()))
        self.tau = tau_initial
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        action_probs = softmax(self.Q[state], self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)  # Set random seed for reproducibility
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            total_reward = 0
            action = self.choose_action(state)

            while not done:
                t, next_state, reward, done = self.gridworld.step(action)
                total_reward += reward
                next_action = self.choose_action(next_state)

                td_target = reward + self.gamma * self.Q[next_state, next_action]
                td_error = td_target - self.Q[state, action]
                self.Q[state, action] += self.alpha * td_error

                state = next_state
                action = next_action
            
            # Track visits and rewards for each goal location
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if state == goal_state:
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            # Store data for plotting
            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            # Decay tau
            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.Q, goal_visits, cumulative_rewards

# Initialize gridworld environments for task 2
gridworld = get_task2_gridworld()

# Run SARSA and Q-learning with different seeds and collect rewards
# seeds = np.random.randint(0, 10000, size=100)
seeds = np.load('seeds.npy')

learning_curves_sarsa = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}
learning_curves_qlearning = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}

for seed in seeds:
    # Q-learning
    ql = QLearningSoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.1, gamma=0.9, episodes=1000)
    Q_ql, _, _ = ql.run(seed=seed)
    learning_curves_qlearning['goal_1_visits'].append(ql.goal_1_visits)
    learning_curves_qlearning['goal_2_visits'].append(ql.goal_2_visits)
    learning_curves_qlearning['goal_1_rewards'].append(ql.goal_1_rewards)
    learning_curves_qlearning['goal_2_rewards'].append(ql.goal_2_rewards)

    # SARSA
    sarsa = SARSASoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.1, gamma=0.9, episodes=1000)
    Q_sarsa, _, _ = sarsa.run(seed=seed)
    learning_curves_sarsa['goal_1_visits'].append(sarsa.goal_1_visits)
    learning_curves_sarsa['goal_2_visits'].append(sarsa.goal_2_visits)
    learning_curves_sarsa['goal_1_rewards'].append(sarsa.goal_1_rewards)
    learning_curves_sarsa['goal_2_rewards'].append(sarsa.goal_2_rewards)

# Average learning curve data across seeds
avg_goal_1_visits_qlearning = np.mean(learning_curves_qlearning['goal_1_visits'], axis=0)
avg_goal_2_visits_qlearning = np.mean(learning_curves_qlearning['goal_2_visits'], axis=0)
avg_goal_1_rewards_qlearning = np.mean(learning_curves_qlearning['goal_1_rewards'], axis=0)
avg_goal_2_rewards_qlearning = np.mean(learning_curves_qlearning['goal_2_rewards'], axis=0)

avg_goal_1_visits_sarsa = np.mean(learning_curves_sarsa['goal_1_visits'], axis=0)
avg_goal_2_visits_sarsa = np.mean(learning_curves_sarsa['goal_2_visits'], axis=0)
avg_goal_1_rewards_sarsa = np.mean(learning_curves_sarsa['goal_1_rewards'], axis=0)
avg_goal_2_rewards_sarsa = np.mean(learning_curves_sarsa['goal_2_rewards'], axis=0)

# Plot the learning curves
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(avg_goal_1_visits_qlearning, label="Q-learning: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_qlearning, label="Q-learning: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("Q-learning: Number of Visits to Each Goal")
plt.legend()
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(avg_goal_1_rewards_qlearning, label="Q-learning: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_qlearning, label="Q-learning: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("Q-learning: Cumulative Rewards for Each Goal")
plt.legend()
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(avg_goal_1_visits_sarsa, label="SARSA: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_sarsa, label="SARSA: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("SARSA: Number of Visits to Each Goal")
plt.legend()
plt.grid()

plt.subplot(2, 2, 4)
plt.plot(avg_goal_1_rewards_sarsa, label="SARSA: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_sarsa, label="SARSA: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("SARSA: Cumulative Rewards for Each Goal")
plt.legend()

plt.grid()

plt.tight_layout()
plt.show()
```

```python
class LinearQFunction:
    def __init__(self, input_size, action_size):
        self.weights = np.random.rand(input_size, action_size)
        
    def predict(self, state):
        return np.dot(state, self.weights)
    
    def update(self, state, action, target, alpha):
        prediction = self.predict(state)[action]
        error = target - prediction
        self.weights[:, action] += alpha * error * state
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def softmax(x, tau):
    """Compute softmax values for each set of scores in x."""
    x = np.array(x) / tau
    x = x - np.max(x)  # For numerical stability
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

class QLearningSoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_function = LinearQFunction(self.state_size, self.action_size)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_function.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state]
            total_reward = 0

            while not done:
                action = self.choose_action(state)
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state]
                total_reward += reward

                best_next_action = np.argmax(self.q_function.predict(next_state))
                td_target = reward + self.gamma * self.q_function.predict(next_state)[best_next_action]
                self.q_function.update(state, action, td_target, self.alpha)

                state = next_state
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state]):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_function, goal_visits, cumulative_rewards

class SARSASoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_function = LinearQFunction(self.state_size, self.action_size)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_function.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state]
            total_reward = 0
            action = self.choose_action(state)

            while not done:
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state]
                total_reward += reward
                next_action = self.choose_action(next_state)


                td_target = reward + self.gamma * self.q_function.predict(next_state)[next_action]
                self.q_function.update(state, action, td_target, self.alpha)

                state = next_state
                action = next_action
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state]):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_function, goal_visits, cumulative_rewards

# Initialize gridworld environments for task 2
gridworld = get_task2_gridworld()

# Run SARSA and Q-learning with linear function approximation using different seeds and collect rewards
seeds = [42, 100, 123, 456, 789]

learning_curves_sarsa = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}
learning_curves_qlearning = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}

for seed in seeds:
    # Q-learning
    ql = QLearningSoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.5, gamma=0.5, episodes=1000)
    Q_ql, _, _ = ql.run(seed=seed)
    learning_curves_qlearning['goal_1_visits'].append(ql.goal_1_visits)
    learning_curves_qlearning['goal_2_visits'].append(ql.goal_2_visits)
    learning_curves_qlearning['goal_1_rewards'].append(ql.goal_1_rewards)
    learning_curves_qlearning['goal_2_rewards'].append(ql.goal_2_rewards)

    # SARSA
    sarsa = SARSASoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.5, gamma=0.5, episodes=1000)
    Q_sarsa, _, _ = sarsa.run(seed=seed)
    learning_curves_sarsa['goal_1_visits'].append(sarsa.goal_1_visits)
    learning_curves_sarsa['goal_2_visits'].append(sarsa.goal_2_visits)
    learning_curves_sarsa['goal_1_rewards'].append(sarsa.goal_1_rewards)
    learning_curves_sarsa['goal_2_rewards'].append(sarsa.goal_2_rewards)

# Average learning curve data across seeds
avg_goal_1_visits_qlearning = np.mean(learning_curves_qlearning['goal_1_visits'], axis=0)
avg_goal_2_visits_qlearning = np.mean(learning_curves_qlearning['goal_2_visits'], axis=0)
avg_goal_1_rewards_qlearning = np.mean(learning_curves_qlearning['goal_1_rewards'], axis=0)
avg_goal_2_rewards_qlearning = np.mean(learning_curves_qlearning['goal_2_rewards'], axis=0)

avg_goal_1_visits_sarsa = np.mean(learning_curves_sarsa['goal_1_visits'], axis=0)
avg_goal_2_visits_sarsa = np.mean(learning_curves_sarsa['goal_2_visits'], axis=0)
avg_goal_1_rewards_sarsa = np.mean(learning_curves_sarsa['goal_1_rewards'], axis=0)
avg_goal_2_rewards_sarsa = np.mean(learning_curves_sarsa['goal_2_rewards'], axis=0)

# Plot the learning curves
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(avg_goal_1_visits_qlearning, label="Q-learning: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_qlearning, label="Q-learning: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("Q-learning: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(avg_goal_1_rewards_qlearning, label="Q-learning: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_qlearning, label="Q-learning: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("Q-learning: Cumulative Rewards for Each Goal")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(avg_goal_1_visits_sarsa, label="SARSA: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_sarsa, label="SARSA: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("SARSA: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(avg_goal_1_rewards_sarsa, label="SARSA: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_sarsa, label="SARSA: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("SARSA: Cumulative Rewards for Each Goal")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
class QLearningSoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_function = LinearQFunction(self.state_size, self.action_size)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_function.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state]
            total_reward = 0

            while not done:
                action = self.choose_action(state)
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state]
                total_reward += reward

                best_next_action = np.argmax(self.q_function.predict(next_state))
                td_target = reward + self.gamma * self.q_function.predict(next_state)[best_next_action]
                self.q_function.update(state, action, td_target, self.alpha)

                state = next_state
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state]):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_function, goal_visits, cumulative_rewards

class SARSASoftmaxDecayingTau:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.alpha = alpha
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_function = LinearQFunction(self.state_size, self.action_size)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_function.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs), p=action_probs)

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state]
            total_reward = 0
            action = self.choose_action(state)

            while not done:
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state]
                total_reward += reward
                next_action = self.choose_action(next_state)

                td_target = reward + self.gamma * self.q_function.predict(next_state)[next_action]
                self.q_function.update(state, action, td_target, self.alpha)

                state = next_state
                action = next_action
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state]):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_function, goal_visits, cumulative_rewards

# Initialize gridworld environments for task 3
gridworld = get_task3_gridworld()

# Run SARSA and Q-learning with linear function approximation using different seeds and collect rewards
seeds = [42, 100, 123, 456, 789]

learning_curves_sarsa = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}
learning_curves_qlearning = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}

for seed in seeds:
    # Q-learning
    ql = QLearningSoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=1000)
    Q_ql, _, _ = ql.run(seed=seed)
    learning_curves_qlearning['goal_1_visits'].append(ql.goal_1_visits)
    learning_curves_qlearning['goal_2_visits'].append(ql.goal_2_visits)
    learning_curves_qlearning['goal_1_rewards'].append(ql.goal_1_rewards)
    learning_curves_qlearning['goal_2_rewards'].append(ql.goal_2_rewards)

    # SARSA
    sarsa = SARSASoftmaxDecayingTau(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, alpha=0.01, gamma=0.9, episodes=1000)
    Q_sarsa, _, _ = sarsa.run(seed=seed)
    learning_curves_sarsa['goal_1_visits'].append(sarsa.goal_1_visits)
    learning_curves_sarsa['goal_2_visits'].append(sarsa.goal_2_visits)
    learning_curves_sarsa['goal_1_rewards'].append(sarsa.goal_1_rewards)
    learning_curves_sarsa['goal_2_rewards'].append(sarsa.goal_2_rewards)

# Average learning curve data across seeds
avg_goal_1_visits_qlearning = np.mean(learning_curves_qlearning['goal_1_visits'], axis=0)
avg_goal_2_visits_qlearning = np.mean(learning_curves_qlearning['goal_2_visits'], axis=0)
avg_goal_1_rewards_qlearning = np.mean(learning_curves_qlearning['goal_1_rewards'], axis=0)
avg_goal_2_rewards_qlearning = np.mean(learning_curves_qlearning['goal_2_rewards'], axis=0)

avg_goal_1_visits_sarsa = np.mean(learning_curves_sarsa['goal_1_visits'], axis=0)
avg_goal_2_visits_sarsa = np.mean(learning_curves_sarsa['goal_2_visits'], axis=0)
avg_goal_1_rewards_sarsa = np.mean(learning_curves_sarsa['goal_1_rewards'], axis=0)
avg_goal_2_rewards_sarsa = np.mean(learning_curves_sarsa['goal_2_rewards'], axis=0)

# Plot the learning curves
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(avg_goal_1_visits_qlearning, label="Q-learning: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_qlearning, label="Q-learning: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("Q-learning: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(avg_goal_1_rewards_qlearning, label="Q-learning: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_qlearning, label="Q-learning: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("Q-learning: Cumulative Rewards for Each Goal")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(avg_goal_1_visits_sarsa, label="SARSA: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_sarsa, label="SARSA: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("SARSA: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(avg_goal_1_rewards_sarsa, label="SARSA: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_sarsa, label="SARSA: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("SARSA: Cumulative Rewards for Each Goal")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np

class SimpleNeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.01):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.learning_rate = learning_rate

        # Initialize weights
        self.weights_input_hidden = np.random.randn(self.input_size, self.hidden_size) * 0.01
        self.weights_hidden_output = np.random.randn(self.hidden_size, self.output_size) * 0.01

    def relu(self, x):
        return np.maximum(0, x)

    def relu_derivative(self, x):
        return np.where(x > 0, 1, 0)

    def predict(self, state):
        self.hidden_layer_input = np.dot(state, self.weights_input_hidden)
        self.hidden_layer_output = self.relu(self.hidden_layer_input)
        output = np.dot(self.hidden_layer_output, self.weights_hidden_output)
        return output

    def update(self, state, target):
        # Forward pass
        self.hidden_layer_input = np.dot(state, self.weights_input_hidden)
        self.hidden_layer_output = self.relu(self.hidden_layer_input)
        output = np.dot(self.hidden_layer_output, self.weights_hidden_output)

        # Compute loss (mean squared error)
        loss = target - output

        # Backward pass
        output_gradient = loss
        hidden_gradient = np.dot(output_gradient, self.weights_hidden_output.T) * self.relu_derivative(self.hidden_layer_input)

        # Update weights
        self.weights_hidden_output += self.learning_rate * np.dot(self.hidden_layer_output.T, output_gradient)
        self.weights_input_hidden += self.learning_rate * np.dot(state.T, hidden_gradient)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(x, tau):
    e_x = np.exp((x - np.max(x)) / tau)
    return e_x / e_x.sum(axis=1, keepdims=True)

class SimpleNeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.01, clip_value=1.0):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.learning_rate = learning_rate
        self.clip_value = clip_value

        # Initialize weights
        self.weights_input_hidden = np.random.randn(self.input_size, self.hidden_size) * 0.01
        self.weights_hidden_output = np.random.randn(self.hidden_size, self.output_size) * 0.01

    def relu(self, x):
        return np.maximum(0, x)

    def relu_derivative(self, x):
        return np.where(x > 0, 1, 0)

    def predict(self, state):
        self.hidden_layer_input = np.dot(state, self.weights_input_hidden)
        self.hidden_layer_output = self.relu(self.hidden_layer_input)
        output = np.dot(self.hidden_layer_output, self.weights_hidden_output)
        return output

    def update(self, state, target):
        # Forward pass
        self.hidden_layer_input = np.dot(state, self.weights_input_hidden)
        self.hidden_layer_output = self.relu(self.hidden_layer_input)
        output = np.dot(self.hidden_layer_output, self.weights_hidden_output)

        # Compute loss (mean squared error)
        loss = target - output

        # Backward pass
        output_gradient = loss
        hidden_gradient = np.dot(output_gradient, self.weights_hidden_output.T) * self.relu_derivative(self.hidden_layer_input)

        # Clip gradients to prevent exploding gradients
        output_gradient = np.clip(output_gradient, -self.clip_value, self.clip_value)
        hidden_gradient = np.clip(hidden_gradient, -self.clip_value, self.clip_value)

        # Update weights
        self.weights_hidden_output += self.learning_rate * np.dot(self.hidden_layer_output.T, output_gradient)
        self.weights_input_hidden += self.learning_rate * np.dot(state.T, hidden_gradient)

class QLearningSoftmaxDecayingTauANN:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, learning_rate=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.learning_rate = learning_rate
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_network = SimpleNeuralNetwork(self.state_size, 24, self.action_size, learning_rate=learning_rate)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_network.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs[0]), p=action_probs[0])

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state].reshape(1, -1)
            total_reward = 0

            while not done:
                action = self.choose_action(state)
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state].reshape(1, -1)
                total_reward += reward

                best_next_action = np.argmax(self.q_network.predict(next_state))
                td_target = reward + self.gamma * self.q_network.predict(next_state)[0][best_next_action]
                target_f = self.q_network.predict(state)
                target_f[0][action] = td_target
                self.q_network.update(state, target_f)

                if np.any(np.isnan(target_f)) or np.any(np.isnan(state)):
                    print("NaN detected in target_f or state")
                    print("State:", state)
                    print("Target_f:", target_f)
                    raise ValueError("NaN detected in target_f or state")

                state = next_state
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state].reshape(1, -1)):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_network, goal_visits, cumulative_rewards

class SARSASoftmaxDecayingTauANN:
    def __init__(self, gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, learning_rate=0.01, gamma=0.9, episodes=100):
        self.gridworld = gridworld
        self.tau_initial = tau_initial
        self.tau_min = tau_min
        self.decay_rate = decay_rate
        self.learning_rate = learning_rate
        self.gamma = gamma
        self.episodes = episodes
        self.tau = tau_initial
        self.state_size = self.gridworld.get_state_size()
        self.action_size = gridworld.get_action_size()
        self.q_network = SimpleNeuralNetwork(self.state_size, 24, self.action_size, learning_rate=learning_rate)
        self.goal_1_visits = []
        self.goal_2_visits = []
        self.goal_1_rewards = []
        self.goal_2_rewards = []

    def choose_action(self, state):
        q_values = self.q_network.predict(state)
        action_probs = softmax(q_values, self.tau)
        if np.any(np.isnan(action_probs)):
            raise ValueError("Action probabilities contain NaN")
        return np.random.choice(len(action_probs[0]), p=action_probs[0])

    def run(self, seed=None):
        np.random.seed(seed)
        goal_visits = {goal: 0 for goal in self.gridworld.get_goal_loc()}
        cumulative_rewards = {goal: 0 for goal in self.gridworld.get_goal_loc()}

        for episode in range(self.episodes):
            t, state, reward, done = self.gridworld.reset()
            state = np.eye(self.gridworld.get_state_size())[state].reshape(1, -1)
            total_reward = 0
            action = self.choose_action(state)

            while not done:
                t, next_state, reward, done = self.gridworld.step(action)
                next_state = np.eye(self.gridworld.get_state_size())[next_state].reshape(1, -1)
                total_reward += reward
                next_action = self.choose_action(next_state)

                td_target = reward + self.gamma * self.q_network.predict(next_state)[0][next_action]
                target_f = self.q_network.predict(state)
                target_f[0][action] = td_target
                self.q_network.update(state, target_f)

                if np.any(np.isnan(target_f)) or np.any(np.isnan(state)):
                    print("NaN detected in target_f or state")
                    print("State:", state)
                    print("Target_f:", target_f)
                    raise ValueError("NaN detected in target_f or state")

                state = next_state
                action = next_action
            
            for goal in self.gridworld.get_goal_loc():
                goal_state = self.gridworld._get_state_from_loc(goal)
                if np.array_equal(state, np.eye(self.gridworld.get_state_size())[goal_state].reshape(1, -1)):
                    goal_visits[goal] += 1
                    cumulative_rewards[goal] += total_reward

            self.goal_1_visits.append(goal_visits[self.gridworld.get_goal_loc()[0]])
            self.goal_2_visits.append(goal_visits[self.gridworld.get_goal_loc()[1]])
            self.goal_1_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[0]])
            self.goal_2_rewards.append(cumulative_rewards[self.gridworld.get_goal_loc()[1]])

            self.tau = max(self.tau_min, self.tau * self.decay_rate)
        
        return self.q_network, goal_visits, cumulative_rewards


# Initialize gridworld environments for task 3
gridworld = get_task3_gridworld()

# Run SARSA and Q-learning with ANN using different seeds and collect rewards
seeds = [42, 100, 123, 456, 789]

learning_curves_sarsa = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}
learning_curves_qlearning = {'goal_1_visits': [], 'goal_2_visits': [], 'goal_1_rewards': [], 'goal_2_rewards': []}

for seed in seeds:
    # Q-learning
    ql = QLearningSoftmaxDecayingTauANN(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, learning_rate=0.5, gamma=0.9, episodes=1000)
    Q_ql, _, _ = ql.run(seed=seed)
    learning_curves_qlearning['goal_1_visits'].append(ql.goal_1_visits)
    learning_curves_qlearning['goal_2_visits'].append(ql.goal_2_visits)
    learning_curves_qlearning['goal_1_rewards'].append(ql.goal_1_rewards)
    learning_curves_qlearning['goal_2_rewards'].append(ql.goal_2_rewards)

    # SARSA
    sarsa = SARSASoftmaxDecayingTauANN(gridworld, tau_initial=1.0, tau_min=0.1, decay_rate=0.99, learning_rate=0.5, gamma=0.9, episodes=1000)
    Q_sarsa, _, _ = sarsa.run(seed=seed)
    learning_curves_sarsa['goal_1_visits'].append(sarsa.goal_1_visits)
    learning_curves_sarsa['goal_2_visits'].append(sarsa.goal_2_visits)
    learning_curves_sarsa['goal_1_rewards'].append(sarsa.goal_1_rewards)
    learning_curves_sarsa['goal_2_rewards'].append(sarsa.goal_2_rewards)

# Average learning curve data across seeds
avg_goal_1_visits_qlearning = np.mean(learning_curves_qlearning['goal_1_visits'], axis=0)
avg_goal_2_visits_qlearning = np.mean(learning_curves_qlearning['goal_2_visits'], axis=0)
avg_goal_1_rewards_qlearning = np.mean(learning_curves_qlearning['goal_1_rewards'], axis=0)
avg_goal_2_rewards_qlearning = np.mean(learning_curves_qlearning['goal_2_rewards'], axis=0)

avg_goal_1_visits_sarsa = np.mean(learning_curves_sarsa['goal_1_visits'], axis=0)
avg_goal_2_visits_sarsa = np.mean(learning_curves_sarsa['goal_2_visits'], axis=0)
avg_goal_1_rewards_sarsa = np.mean(learning_curves_sarsa['goal_1_rewards'], axis=0)
avg_goal_2_rewards_sarsa = np.mean(learning_curves_sarsa['goal_2_rewards'], axis=0)

# Plot the learning curves
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(avg_goal_1_visits_qlearning, label="Q-learning: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_qlearning, label="Q-learning: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("Q-learning: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(avg_goal_1_rewards_qlearning, label="Q-learning: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_qlearning, label="Q-learning: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("Q-learning: Cumulative Rewards for Each Goal")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(avg_goal_1_visits_sarsa, label="SARSA: Goal (2, 0) Visits")
plt.plot(avg_goal_2_visits_sarsa, label="SARSA: Goal (6, 9) Visits")
plt.xlabel("Episodes")
plt.ylabel("Number of Visits")
plt.title("SARSA: Number of Visits to Each Goal")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(avg_goal_1_rewards_sarsa, label="SARSA: Goal (2, 0) Rewards")
plt.plot(avg_goal_2_rewards_sarsa, label="SARSA: Goal (6, 9) Rewards")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Rewards")
plt.title("SARSA: Cumulative Rewards for Each Goal")
plt.legend()

plt.grid()
plt.tight_layout()
plt.show()
```
