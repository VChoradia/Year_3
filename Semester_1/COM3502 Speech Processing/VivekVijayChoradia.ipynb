```python
def design_filter(filter_type, cutoff, fs, order=10):
    """
    Designs a Butterworth filter.

    Parameters:
    filter_type (str): The type of filter ('low', 'high').
    cutoff (float or list): The cutoff frequency for the filter.
    fs (int): The sampling frequency.
    order (int, optional): The order of the filter. Defaults to 10.

    Returns:
    tuple: Numerator (b) and denominator (a) polynomials of the IIR filter.
    """
    b, a = sig.butter(order, cutoff, fs=fs, btype=filter_type, analog=False)
    return b, a

def design_band_filter(filter_type, low, high, fs, order=10):
    """
    Designs a band-stop/band-pass Butterworth filter.

    Parameters are the same as the `design_filter` function but with separate low and high cutoff frequencies.

    Returns:
    tuple: Numerator (b) and denominator (a) polynomials of the IIR filter.
    """
        
    b, a = sig.butter(order, [low, high], fs=fs, btype=filter_type, analog=False)
    return b, a

def visualise_filter(b, a, fs, cutoff, filter, xlim=1500, title=None):
    """
    Visualizes the frequency response and pole-zero plot of a filter.

    Parameters:
    b, a (array): Numerator and denominator polynomials of the IIR filter.
    fs (int): The sampling frequency.
    cutoff (float or list): The cutoff frequency(ies).
    filter (str): Type of filter for specific plot adjustments.
    xlim (int, optional): x-axis limit for the frequency response plot. Defaults to 1500.
    title (list of str, optional): Titles for the subplots. Defaults to None.
    """
    
    w, h = sig.freqz(b, a, fs=fs, worN=8000)
    
    plt.figure(figsize=(15, 6))
    
    plt.subplot(1, 2, 1)
    plt.plot(w, np.abs(h))
    plt.xlim(0,xlim)
    
    if filter not in ['bandstop','band']:
        plt.plot(cutoff, 0.5*np.sqrt(2), 'ko-')
        plt.axvline(cutoff, color='red')
    else:
        plt.axvline(cutoff[0], color='red')
        plt.axvline(cutoff[1], color='red')
    

    plt.title(title[0])
    plt.xlabel('Frequency [Hz]')
    plt.ylabel('Amplitude [dB]')

    
    plt.subplot(1, 2, 2)
    z, p, k = sig.tf2zpk(b, a)
    
    z, p = np.roots(z), np.roots(p)
    
    ax = plt.gca()
    ax.plot(np.real(z), np.imag(z), 'bo', fillstyle='none', ms=10)
    ax.plot(np.real(p), np.imag(p), 'rx', fillstyle='none', ms=10)
    unit_circle = plt.Circle((0, 0), radius=1, fill=False,
                             color='black', ls='--', alpha=0.9)
    ax.add_patch(unit_circle)

    plt.title(title[1])
    plt.xlabel('Re{$z$}')
    plt.ylabel('Im{$z$}')
    plt.axis('equal')

    
    plt.tight_layout()

    plt.show()
    
```

```python
# Design a high-pass filter with a cut-off frequency of ≈500Hz using a filter design method of your choice.
b_hp, a_hp = design_filter('high', 500, fs, order=6)
visualise_filter(b_hp, a_hp, fs, 500, 'high', title=['High Pass Filter Function', 'High Pass Filter Zero-Point Plot'])

high_pass_audio = sig.lfilter(b_hp, a_hp, s)
ipd.Audio(high_pass_audio, rate=fs)
```

```python
# Design a low-pass filter with a cut-off frequency of  ≈500 Hz.
b_lp, a_lp = design_filter('low', 500, fs, order=6)
visualise_filter(b_lp, a_lp, fs, 500, 'low', title=['Low Pass Filter Function', 'Low Pass Filter Zero-Point Plot'])

low_pass_audio = sig.lfilter(b_lp, a_lp, s)
ipd.Audio(low_pass_audio, rate=fs)
```

```python
# Design a band-stop filter with a cut-off frequencies of  ≈300Hz and of  ≈1.1kHz.
low = 300
high = 1100
b_bs, a_bs = design_band_filter('bandstop', low, high, fs, order=5)
visualise_filter(b_bs, a_bs, fs, [low, high], 'bandstop', title=['Band Stop Filter Function', 'Band Stop Filter Zero-Point Plot'])

band_stop_audio = sig.lfilter(b_bs, a_bs, s)
ipd.Audio(band_stop_audio, rate=fs)
```

### 4.1 Low-Frequency Oscillator

Many ‘Voice effects (FXs)’ are achieved by modifying some characteristic of the speech using a low-frequency oscillator or *LFO*. LFOs typically have two controls: speed (which is specified by the frequency in Hertz) and depth (which specifies the magnitude of the effect). The following tasks will require several LFOs, so it makes sense to implement one in the following.

<br>
<div style="border: 2px solid #999; padding: 10px; background: #abe;">
    
**Task T3:**
    
* Implement a Low Frequency Oscillator as a function `lfo()` as described below. Visualise that your function works by generating a sine and a square wave of frequency $5$ Hz and length $2$ seconds with different depths.
    
Note: There will be an extra point in the marking if you **do not** use the `scipy` library to solve this task.
    
</div>

```python
def lfo(speed_hz, depth, num_samples, fs=44100, square_curve=False):
    '''
    Low-frequency oscillator
    
    Parameters
    ----------
    speed_hz : float
       frequency of generated signal in Hertz
    depth : float
        magnitude of the effect
    num_samples : int
        length of the signal in samples
    fs : float, optional 
        sampling frequency in Hz, default 44100
    square_curve : boolean, optional (default: False)
        generate square wave if true, generate sine wave if false

    Example use:
    -------
        sig_square = lfo(speed_hz=5, depth=0.7, num_samples=88200, fs=44100, square_curve=True)
    '''
    
    t = np.arange(num_samples) / fs
    if square_curve:
        # Square wave generation 
        x = depth * np.sign(np.sin(2 * np.pi * speed_hz * t))
    else:
        # Sine wave generation
        x = depth * np.sin(2 * np.pi * speed_hz * t)
    
    return x
```

```python
def visualise_lfo(time_vector, sine_wave, square_wave, depth):
    """
    Visualizes sine and square waveforms typically used in LFOs.

    Parameters
    ----------
    time_vector (array):Time values.
    sine_wave (array): An array representing the sine wave at given time points.
    square_wave (array): An array representing the square wave at given time points.
    depth (float): The depth of the LFO, affecting the amplitude of the waveforms.

    Returns:
    None: This function plots the waveforms but does not return any value.
    """
    
    # Plot the generated waves
    plt.figure(figsize=(18, 5))

    # Sine wave plot
    plt.subplot(1, 2, 1)
    plt.plot(time_vector, sine_wave, label=f"5Hz Sine Wave of length 2 seconds and depth {str(depth)}")
    plt.title('5 Hz Sine Wave')
    plt.xlabel('Time [s]')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.legend()

    # Square wave plot
    plt.subplot(1, 2, 2)
    plt.plot(time_vector, square_wave, label=f"5Hz Square Wave of length 2 seconds and depth {str(depth)}", color='orange')
    plt.title('5 Hz Square Wave')
    plt.xlabel('Time [s]')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python
# Code to show that the LFO works

num_samples = 88200
time_vector = np.arange(num_samples)/fs

# Generate a 5Hz sine wave and a square wave, both with length of 2 seconds and depth of 1.0
sine_wave = lfo(speed_hz=5, num_samples=num_samples, depth=1.0, square_curve=False)
square_wave = lfo(speed_hz=5,num_samples=88200, depth=1.0, square_curve=True)
visualise_lfo(time_vector, sine_wave, square_wave,1.0)

# Generate a 5Hz sine wave and a square wave, both with length of 2 seconds and depth of 0.5
sine_wave = lfo(speed_hz=5, num_samples=num_samples, depth=0.5, square_curve=False)
square_wave = lfo(speed_hz=5,num_samples=88200, depth=0.5, square_curve=True)
visualise_lfo(time_vector, sine_wave, square_wave,0.5)

# Generate a 5Hz sine wave and a square wave, both with length of 2 seconds and depth of 0.2
sine_wave = lfo(speed_hz=5, num_samples=num_samples, depth=0.2, square_curve=False)
square_wave = lfo(speed_hz=5,num_samples=88200, depth=0.2, square_curve=True)
visualise_lfo(time_vector, sine_wave, square_wave,0.2)
```

### 4.2 Amplitude Modulation - Tremolo

*Tremolo* is one of the most basic voice manipulations that makes use of an LFO. In this effect, the amplitude of a speech signal is [modulated](https://en.wikipedia.org/wiki/Amplitude_modulation), i.e. the speech waveform is multiplied by a variable gain that ranges between $0$ and $1$. 

Your LFO outputs an audio signal between `-depth` and `+depth`. So, in order to modulate the amplitude of the speech correctly, the output of the LFO has to be scaled appropriately to range between $0$ and $1$.

<br>
<div style="border: 2px solid #999; padding: 10px; background: #abe;">
    
**Task T4:**
    
* Implement  a function `tremolo()` using your function `lfo()` and modulate the amplitude of the speech signal. 
* Experiment with different settings for `speed` and `depth`. In particular, note that a square wave with a *speed* between $3$ and $4$ Hz (and depth = $1$) has a very destructive effect on the intelligibility of the output. This is because $3-4$ Hz corresponds to the typical syllabic rate of speech.
* Proof that the effect works by a proper visualisation of the filtered speech signal and describe what can be observed and perceived.
    
</div>

```python
def tremolo(signal, fs, speed_hz, depth, square_curve=False):
    '''
    Applies a tremolo effect to a signal
    
    Parameters
    ----------
    signal : float
       input signal to which the effect should be applied
    fs : int
        sampling frequency in Hz
    speed_hz : float
       frequency of LFO and by this also the effect
    depth : float
        magnitude of the effect
    square_curve : boolean, optional
        generate square wave if true, generate sine wave if false
    
    Return
    ----------
    signal after application of tremolo effect
    
    Example use:
    -------
        signal_tremolo = tremolo(audio_in, fs, 10, 1, square_curve=False)
    '''

    num_samples = len(signal)
    lfo_signal = lfo(speed_hz, depth, num_samples, fs, square_curve)
    
    # Normalize Lfo signal to range (0,1)
    normalized_lfo_signal = (lfo_signal + depth) / (2 * depth)
    
    # Apply tremolo effect to the signal
    modulated_signal = normalized_lfo_signal * signal
    
    return modulated_signal
```

```python
def visualise_comparision_graphs(signal, signal_tremolo, zoom=False):
    """
    Visualizes and compares an original audio signal with its tremolo-processed version.

    Parameters:
    -----------
    signal (array): The original audio signal.
    signal_tremolo (array): The tremolo-processed audio signal.
    zoom (bool, optional): If True, zooms into a specific part of the graph. Defaults to False.
    """
    
    num_samples = len(signal)
    time_vector = np.arange(num_samples) / fs
    
    #Plot the graphs
    fig=plt.figure(figsize=(20,5))
    
    # Plotting the original sound file
    plt.subplot(1,3,1)
    plt.title('Original Sound File')
    plt.xlabel('Time t[s]')
    plt.ylabel('Amplitude')
    plt.plot(time_vector, s, label="Amplitude over time", color='red')
    plt.grid()
    plt.legend()  
    if zoom: 
        plt.title('Original Sound File (Zoomed)')
        plt.xlim(0.5,2.1)
   
    # Plotting the output (tremolo-processed) sound file
    plt.subplot(1,3,2)
    plt.title('Output Sound File')
    plt.xlabel('Time t[s]')
    plt.ylabel('Amplitude')
    plt.plot(time_vector, signal_tremolo, color='blue', label="Amplitude over time")
    plt.grid()
    plt.legend()  
    if zoom: 
        plt.title('Output Sound File (Zoomed)')
        plt.xlim(0.5,2.1)
    
    # Plotting the difference between the original and output sound files
    plt.subplot(1,3,3)
    plt.title('Difference in Original and Output Sound Files')
    plt.xlabel('Time t[s]')
    plt.ylabel('Amplitude')
    plt.plot(time_vector, signal - signal_tremolo, color="green", label="Amplitude over time")
    plt.grid()
    plt.legend()  
    if zoom: 
        plt.title('Difference in Original and Output Sound Files (Zoomed)')
        plt.xlim(0.5,2.1)
        
     
    plt.show()
    
```

```python
print("Audio after tremolo with LFO frequency: 3Hz, depth: 1 with Square Wave:")
signal_tremolo = tremolo(s, fs, speed_hz=3, depth=1, square_curve=True)
visualise_comparision_graphs(s, signal_tremolo)
visualise_comparision_graphs(s, signal_tremolo, zoom=True)
ipd.Audio(signal_tremolo, rate = fs)
```

```python
print("Audio after tremolo with LFO frequency: 15Hz, depth: 5 with Sine Wave :")
signal_tremolo = tremolo(s, fs, speed_hz=15, depth=0.5, square_curve=False)
visualise_comparision_graphs(s, signal_tremolo)
visualise_comparision_graphs(s, signal_tremolo, zoom=True)
ipd.Audio(signal_tremolo, rate = fs)
```

```python
print("Audio after tremolo with LFO frequency: 100Hz, depth: 12 with Sine Wave :")
signal_tremolo = tremolo(s, fs, speed_hz=100, depth=12, square_curve=False)
visualise_comparision_graphs(s, signal_tremolo)
visualise_comparision_graphs(s, signal_tremolo, zoom=True)
ipd.Audio(signal_tremolo, rate = fs)
```

### 4.4 Frequency Shifting

Many Vocal FX are the result of altering the frequencies present, e.g. changing the pitch of a voice. There are many algorithms for frequency shifting. You have already implemented an approximate solution with your ring modulator.

For simplicity, the following function will be given implementing frequency shifting.

```python
# the following code in this cell is taken and slightly adapted from:
# https://gist.github.com/lebedov/4428122

import scipy.signal as sig

def nextpow2(n):
    '''Return the first integer N such that 2**N >= abs(n)'''
    return int(np.ceil(np.log2(np.abs(n))))

def frequency_shift(signal, fs, shift_amount):
    '''
    Shift the specified signal by the specified frequency.
    
    Parameters
    ----------
    signal : float
       input signal to which the effect should be applied
    fs : int
        sampling frequency in Hz
    shift_amount : float
       amount of frequency shift (in Hz)
   
    Return
    ----------
    signal after application of the frequency shifting effect
    
    Example use:
    -------
        signal_frequency_shifted = frequency_shift(audio_in, fs, 100)
    '''

    # Pad the signal with zeros to prevent the FFT invoked by the transform from
    # slowing down the computation:
    N_orig = len(signal)
    N_padded = 2 ** nextpow2(N_orig)
    t = np.arange(0, N_padded)
    return (
        sig.hilbert(
            np.hstack((signal, np.zeros(N_padded - N_orig, signal.dtype)))
        )
        * np.exp(2j * np.pi * shift_amount * t / fs)
    )[:N_orig].real
```

### 4.5 Harmony Effect

A classic ‘robotic’ voice can be achieved by simply adding frequency-shifted speech back to the unprocessed original. This effect is known as ‘harmony’. However, rather than simply adding the signals in equal amounts, we will implement a more general-purpose approach.

```python
def mixer(signal1, signal2, percentage_l=0.5):
    """
    Mixes the given 2 signals as per the given percentage
    """
    mix_signal = ((1-percentage_l)*signal1) + (percentage_l*signal2)
    return mix_signal
```

```python
def harmony(original_signal, fs, shift_amount, percentage_l=0.5):
    """
    Creates a harmony effect by frequency-shifting the original signal and mixing it with the original.

    Parameters:
    original_signal (array): The original audio signal array.
    fs (int): The sampling frequency of the audio signal.
    shift_amount (float): The amount by which the frequency of the original signal should be shifted.
    percentage_l (float, optional): The mix ratio between the original and shifted signals. 
                                    Defaults to 0.5, meaning equal mix of both.

    Returns:
    mixed_signal (array): The resultant audio signal after applying the harmony effect.
    """
    
    frequency_shifted_signal = frequency_shift(original_signal, fs, shift_amount)
    mixed_signal = mixer(original_signal, frequency_shifted_signal, percentage_l)
    
    return mixed_signal
```

  * below $20$ msecs
  * between $20$ and $100$ msecs, and

```python
def echo(signal, delay, fs, echo_volume=0.5):
    '''
    Applies an echo effect to a signal.

    Parameters
    ----------
    signal : ndarray
        The input signal to which the effect should be applied.
    delay : float
        The delay time in seconds.
    fs : int
        The sampling frequency in Hz.
    echo_volume : float
        The volume of the echo relative to the original signal (0 to 1).

    Returns
    -------
    ndarray
        The signal with an echo effect.

    Example use:
    -------
        signal_echo = echo(audio_in, 0.1, fs, 0.5)
    '''
    # Calculate the number of delay samples
    delay_samples = int(delay * fs)

    # Create a delayed version of the signal
    delayed_signal = np.zeros_like(signal)
    if delay_samples < len(signal):
        delayed_signal[delay_samples:] = signal[:-delay_samples]

    # Mix the original signal with the delayed signal
    output_signal = signal + delayed_signal * echo_volume

    return output_signal
```

### 4.9 Flanger 

It is possible to use an LFO to vary the delay. The resulting effect is known as a *flanger*.

<br>
<div style="border: 2px solid #999; padding: 10px; background: #abe;">
    
**Task T11:**
    
* Add an LFO to your ‘delay’ to create a ‘flanger’, and experiment with different settings. Note that you will need to scale the output of the LFO, and you will get different effects depending on whether the delayed signal is mixed with the original or not.
</div>

```python
def apply_flanger_effect(audio_input, delay, fs, speed_hz, depth, square_wave):
    """
    Implements a flanger effect on an audio input.

    Parameters:
    audio_input (ndarray): The original audio signal.
    delay (float): Base delay time in seconds for the flanger effect.
    fs (int): Sampling rate of the audio signal.
    speed_hz (float): Rate at which the flanger effect oscillates.
    depth (float): Intensity of the effect, ranging from 0 to 1.
    square_wave (bool): Determines if the modulation waveform is square.

    Returns:
    ndarray: The audio signal with the flanger effect applied.
    """

    signal_length = len(audio_input)
    audio_input = np.asarray(audio_input)

    # Convert delay and depth from time to samples
    delay_in_samples = int(delay * fs)
    depth_in_samples = int(depth * fs)

    # Create the Low-Frequency Oscillator (LFO) signal
    lfo_signal = lfo(speed_hz, depth, signal_length, fs, square_wave)
    lfo_signal = (lfo_signal + 1) / 2 * depth_in_samples
    lfo_signal = lfo_signal.astype(int)

    # Implement the flanger effect by adding the delayed signal
    flanged_signal = np.zeros_like(audio_input)
    for index in range(signal_length):
        delay_index = index - delay_in_samples - lfo_signal[index]
        flanged_signal[index] = audio_input[index] + audio_input[delay_index]

    return flanged_signal
```

The following function realises one of the sliders in software.

```python
def peaking_filter(gain,center_freq,q,fs):
    """
    Derive coefficients for a peaking filter with a given amplitude and
     bandwidth.  All coefficients are calculated as described in Zölzer's
     DAFX book (ISBN: 0-471-49078-4, p. 50 - 55).  This algorithm assumes 
     a constant q-term is used through the equation.
    
    Usage:     `b,a` = peaking_filter(gain,center_freq, q,fs)
                `gain` is the logarithmic gain (in dB)
                `center_freq` is the center frequency
                `q` is q-term equating to (Fb / Fc)
                `fs` is the sampling rate
    
    Author:    Jeff Tackett 08/22/05
    Port to Python by George Close 10/07/21
    """
    
    gain = np.float32(gain)
    k = np.tan((np.pi*center_freq)/fs)
    V0 = 10**((gain)/20)
    # invert gain if a cut
    if V0 < 1:
        V0 = 1/V0

    # Boost
    if gain > 0:
        b0 = (1 + ((V0/q)*k)+ k**2) / (1+((1/q)*k)+k**2)
        b1 = (2 * (k**2 - 1)) / (1 + ((1/q)*k) + k**2);
        b2 = (1 - ((V0/q)*k) + k**2) / (1 + ((1/q)*k) + k**2);
        a1 = b1;
        a2 =  (1 - ((1/q)*k) + k**2) / (1 + ((1/q)*k) + k**2);
    # Cut
    elif gain <0:
        b0 = (1 + ((1/q)*k) + k**2) / (1 + ((V0/q)*k) + k**2);
        b1 =       (2 * (k**2 - 1)) / (1 + ((V0/q)*k) + k**2);
        b2 = (1 - ((1/q)*k) + k**2) / (1 + ((V0/q)*k) + k**2);
        a1 = b1;
        a2 = (1 - ((V0/q)*k) + k**2) / (1 + ((V0/q)*k) + k**2);
    #gain is 0
    else:
        b0 = V0;
        b1 = 0;
        b2 = 0;
        a1 = 0;
        a2 = 0;
    a = [  1, a1, a2];
    b = [ b0, b1, b2];
    return b,a
```
